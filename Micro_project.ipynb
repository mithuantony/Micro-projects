# 2. Introduction

Quantum confinement describes the effect that occurs when a particle such as
an electron is restricted to a very small region of space.

For nanoscale electronic structures, classical physics is not sufficient to
describe the behaviour of electrons. The Schrödinger equation provides a
mathematical model for their quantum behaviour.

In this project, an electron is considered inside a one-dimensional infinite
potential well of length L.

The potential is

V(x) = 0,       0 < x < L

V(x) = ∞,       outside the well.

The one-dimensional time-dependent Schrödinger equation is

$iℏ ∂Ψ(x,t)/∂t =
−(ℏ²/2m) ∂²Ψ(x,t)/∂x²$

with boundary conditions

$Ψ(0,t) = 0$

$Ψ(L,t) = 0.$

# 3. Separation of Variables

Assume that the wave function can be written as

$Ψ(x,t) = ψ(x) φ(t)$

Substituting this into the Schrödinger equation,

$iℏ ψ(x) dφ/dt =
−(ℏ²/2m) φ(t) d²ψ/dx²$

Dividing by $ψ(x)φ(t)$,

$iℏ (1/φ) dφ/dt =
−(ℏ²/2m)(1/ψ) d²ψ/dx²$

The left-hand side depends only on time and the right-hand side depends only
on position. Therefore, both sides must be equal to a constant.

Let the separation constant be E.

Therefore,

$iℏ dφ/dt = Eφ$

and

$−(ℏ²/2m)d²ψ/dx² = Eψ.$

# 4. Time-Dependent Solution

From

$iℏ dφ/dt = Eφ$

we get

$dφ/dt = −iE/ℏ φ$

Solving,

$φ(t) = exp(−iEt/ℏ)$

Therefore, the time-dependent part of the wave function is

$φ(t) = e^(−iEt/ℏ^)$.

# 5. Spatial Solution

The spatial equation is

$−(ℏ²/2m)d²ψ/dx² = Eψ$

Rearranging,

$d²ψ/dx² + k²ψ = 0$

where

$k² = 2mE/ℏ².$

The general solution is

$ψ(x) = A sin(kx) + B cos(kx)$.

# 6. Applying Boundary Conditions

The first boundary condition is

$ψ(0) = 0$.

Therefore,

$A sin(0) + B cos(0) = 0$

B = 0.

Hence,

$ψ(x) = A sin(kx)$.

The second boundary condition is

$ψ(L) = 0$.

Therefore,

A sin(kL) = 0.

For a non-zero wave function,

sin(kL) = 0.

Hence,

kL = nπ

where

n = 1, 2, 3, ...

Therefore,

k_n = nπ/L.

# 7. Quantized Energy Levels

We know that

$k² = 2mE/ℏ²$.

Therefore,

$E = ℏ²k²/(2m)$.

Substituting

$k_n = nπ/L$,

we obtain

$E_n = n²π²ℏ²/(2mL²)$.

Therefore, the allowed energy states are

 $E_n = n²π²ℏ² / (2mL²)$               

 where n = 1, 2, 3, ...

This shows that the energy of the confined electron is quantized.

# 8. Normalized Wave Function

The spatial wave function is

$ψ_n(x) = A sin(nπx/L)$.

For normalization,

$∫₀ᴸ |ψ_n(x)|² dx = 1.$

Using

$∫₀ᴸ sin²(nπx/L) dx = L/2,$

we obtain

$A = √(2/L)$.

Therefore,

$ψ_n(x) = √(2/L) sin(nπx/L)$.

Including the time-dependent part,

$Ψ_n(x,t) =
√(2/L) sin(nπx/L) e^(−iE_nt/ℏ^).$

This is the normalized stationary-state wave function.

# 9. General Fourier-Series Solution

A general wave function can be represented as a Fourier series:

$Ψ(x,t) =
Σ c_n √(2/L) sin(nπx/L)e^(−iE_nt/ℏ)$

where

$c_n =
√(2/L) ∫₀ᴸ Ψ(x,0)
sin(nπx/L) dx.$

The coefficients c_n determine the contribution of each energy state to the
initial wave function.

# 10. Probability Density

The probability density of finding the particle at position x is

$P(x,t) = |Ψ(x,t)|²$.

For a stationary state,

$Ψ_n(x,t) =
ψ_n(x)e^(−iE_nt/ℏ^)$.

Since

$|e^(−iE_nt/ℏ^)|² = 1$,

we get

$|Ψ_n(x,t)|² = |ψ_n(x)|²$.

Therefore,

$|Ψ_n(x,t)|² =
(2/L) sin²(nπx/L)$.

The probability density is therefore independent of time for an individual
stationary state.

```python
import matplotlib.pyplot as plt
import numpy as np

# Width of the infinite potential well
L = 5e-9                     # 5 nm

# Position grid
x = np.linspace(0, L, 1000)

# First three energy levels
n_values = [1, 2, 3]

# Function for normalized spatial wave function
def psi(n, x):
    return np.sqrt(2/L) * np.sin(n*np.pi*x/L)

# Probability density
def probability_density(n, x):
    wave = psi(n, x)
    return np.abs(wave)**2


plt.figure(figsize=(10,6))

for n in n_values:
    probability = probability_density(n, x)

    # Multiply by L to make the graph dimensionless
    plt.plot(
        x*1e9,
        probability*L,
        label=f"n = {n}"
    )

plt.xlabel("Position x (nm)")
plt.ylabel(r"$L|\Psi(x,t)|^2$")
plt.title("Probability Density for First Three Energy Levels")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Check normalization numerically

print("Normalization Check")
print("-------------------")

for n in n_values:

    probability = probability_density(n, x)

    integral = np.trapezoid(
        probability,
        x
    )

    print(
        f"n = {n} : "
        f"∫|Ψ|² dx = {integral:.8f}"
    )
```

# 11. Interpretation of the Probability Density

### Ground State: n = 1

The ground state has no internal node. The probability density has a single
maximum near the centre of the well.

### First Excited State: n = 2

The n = 2 state has one internal node. The probability density therefore has
two regions of high probability.

### Second Excited State: n = 3

The n = 3 state has two internal nodes and three regions of high probability.

As n increases, the number of nodes increases.

The energy increases according to

E_n ∝ n².

```python
import numpy as np
import matplotlib.pyplot as plt

# Physical constants (copied from fumT-pAh-in7 for self-containment)
hbar = 1.054571817e-34       # Reduced Planck constant (J s)
m = 9.1093837e-31            # Electron mass (kg)
e = 1.602176634e-19          # Electron charge (C)

# Width of the infinite potential well (copied from fumT-pAh-in7)
L = 5e-9                     # 5 nm

# Position grid (copied from fumT-pAh-in7)
x = np.linspace(0, L, 1000)

# Function to calculate energy (copied from lkNVUGTh-q8U)
def energy(n):
    return (n**2 * np.pi**2 * hbar**2) / (2 * m * L**2)

# Function for normalized spatial wave function (copied from 9IcGDLAW-1_E)
def psi(n, x):
    return np.sqrt(2/L) * np.sin(n*np.pi*x/L)


# Time-dependent wave function

def time_factor(n, t):
    return np.exp(-1j * energy(n) * t / hbar)


def wave_function(n, x, t):
    return psi(n, x) * time_factor(n, t)


# Times in seconds
times = [
    0,
    0.25e-15,
    0.50e-15
]

n = 1

plt.figure(figsize=(10,6))

for t in times:

    wave = wave_function(n, x, t)

    plt.plot(
        x*1e9,
        np.real(wave) / np.max(np.abs(wave)),
        label=f"t = {t*1e15:.2f} fs"
    )

plt.xlabel("Position x (nm)")
plt.ylabel("Normalized Re[Ψ]")
plt.title("Time Evolution of the n = 1 Wave Function")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Show that probability density remains unchanged with time
import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))

for t in times:

    wave = wave_function(n, x, t)
    probability = np.abs(wave)**2

    plt.plot(
        x*1e9,
        probability*L,
        label=f"t = {t*1e15:.2f} fs"
    )

plt.xlabel("Position x (nm)")
plt.ylabel(r"$L|\Psi(x,t)|^2$")
plt.title("Probability Density at Different Times")

plt.grid(True)
plt.legend()

plt.show()
```
